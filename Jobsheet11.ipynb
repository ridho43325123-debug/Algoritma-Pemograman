Pratikum 01 Sequential Search pada Data Angka

```python
data = [10, 20, 30, 40, 50]
target = 30

posisi = -1
for i in range(len(data)):
    if data[i] == target:
        posisi = i
        break

if posisi != -1:
    print("Data ditemukan di indeks", posisi)
else:
    print("Data tidak ditemukan")
```

Pratikum 02 Binary Search pada Data Angka

```python
data = [10, 20, 30, 40, 50]  # sudah terurut
target = 40

low = 0
high = len(data) - 1
posisi = -1

while low <= high:
    mid = (low + high) // 2
    if data[mid] == target:
        posisi = mid
        break
    elif data[mid] < target:
        low = mid + 1
    else:
        high = mid - 1

if posisi != -1:
    print("Data ditemukan di indeks", posisi)
else:
    print("Data tidak ditemukan")
```

Pratikum 05 Perbandingan Waktu Sequential vs Binary Search

```python
import time

# data 1..100000
data = list(range(1, 100001))
target = 99999

# Sequential search
start = time.time()
posisi_seq = -1
for i in range(len(data)):
    if data[i] == target:
        posisi_seq = i
        break
waktu_seq = time.time() - start

# Binary search
low = 0
high = len(data) - 1
posisi_bin = -1

start = time.time()
while low <= high:
    mid = (low + high) // 2
    if data[mid] == target:
        posisi_bin = mid
        break
    elif data[mid] < target:
        low = mid + 1
    else:
        high = mid - 1
waktu_bin = time.time() - start

print("Sequential:", posisi_seq, "waktu =", waktu_seq)
print("Binary    :", posisi_bin, "waktu =", waktu_bin)
```

Penugasan 01 Variasi Sequential Search (hitung perbandingan)


```python
def sequential_search_variasi(data, target):
    comparisons = 0
    for i, x in enumerate(data):
        comparisons += 1
        if x == target:
            return i, comparisons
    return -1, comparisons

data = [10, 20, 30, 40, 50]
target = 30
pos, comp = sequential_search_variasi(data, target)
print("Posisi:", pos, "Perbandingan:", comp)
```

Penugasan 02 Variasi Binary Search (hitung langkah/iterasi)

```python
def binary_search_variasi(data, target):
    low, high = 0, len(data) - 1
    steps = 0
    while low <= high:
        steps += 1
        mid = (low + high) // 2
        if data[mid] == target:
            return mid, steps
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, steps

data = [10, 20, 30, 40, 50]
target = 40
pos, steps = binary_search_variasi(data, target)
print("Posisi:", pos, "Langkah:", steps)
```

Penugasan 04 Studi Kasus Barang Toko

```python
# contoh data barang (nama, harga)
barang = [
    ("Sabun", 5000),
    ("Sampo", 15000),
    ("Pasta Gigi", 12000),
    ("Tisu", 8000),
]

# urutkan berdasarkan harga
barang.sort(key=lambda x: x[1])

target_harga = int(input("Masukkan harga yang dicari: "))

low, high = 0, len(barang) - 1
posisi = -1
while low <= high:
    mid = (low + high) // 2
    if barang[mid][1] == target_harga:
        posisi = mid
        break
    elif barang[mid][1] < target_harga:
        low = mid + 1
    else:
        high = mid - 1

if posisi != -1:
    print("Barang dengan harga", target_harga, "adalah", barang[posisi][0])
else:
    print("Barang dengan harga tersebut tidak ditemukan")
```

Penugasan 05 Analisis Kompleksitas

```python
import time

def seq_search(data, target):
    for x in data:
        if x == target:
            return True
    return False

def bin_search(data, target):
    low, high = 0, len(data) - 1
    while low <= high:
        mid = (low + high) // 2
        if data[mid] == target:
            return True
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return False

ukuran_data = [100, 1000, 10000]
target = -1  # supaya kasus terburuk

hasil = []
for n in ukuran_data:
    data = list(range(n))

    start = time.time()
    seq_search(data, target)
    t_seq = time.time() - start

    start = time.time()
    bin_search(data, target)
    t_bin = time.time() - start

    hasil.append((n, t_seq, t_bin))

for n, t_seq, t_bin in hasil:
    print(f"n={n}: sequential={t_seq:.6f}s, binary={t_bin:.6f}s")
```

Penugasan 06 Proyek Mini Pencarian Nilai Mahasiswa

```python
# input daftar (nama, nilai)
banyak = int(input("Masukkan jumlah mahasiswa: "))
mahasiswa = []

for i in range(banyak):
    nama = input(f"Nama mahasiswa ke-{i+1}: ")
    nilai = int(input(f"Nilai {nama}: "))
    mahasiswa.append((nama, nilai))

# urutkan berdasarkan nilai
mahasiswa.sort(key=lambda x: x[1])

print("\nDaftar mahasiswa terurut (berdasarkan nilai):")
for m in mahasiswa:
    print(m[0], m[1])

# cari nilai tertentu dengan binary search
target_nilai = int(input("\nMasukkan nilai yang ingin dicari: "))

low, high = 0, len(mahasiswa) - 1
posisi = -1
while low <= high:
    mid = (low + high) // 2
    if mahasiswa[mid][1] == target_nilai:
        posisi = mid
        break
    elif mahasiswa[mid][1] < target_nilai:
        low = mid + 1
    else:
        high = mid - 1

if posisi != -1:
    print("Nilai ditemukan pada mahasiswa:", mahasiswa[posisi][0])
else:
    print("Tidak ada mahasiswa dengan nilai tersebut")
```
