Pratikum 01 Bubble sort


```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n-1):
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

data1 = [64, 34, 25, 12, 22, 11, 90]
print("Output Bubble Sort :", bubble_sort(data1))
```

Pratikum 02 Selection sort

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

data2 = [64, 25, 12, 22, 11]
print("Output Selection Sort :", selection_sort(data2))
```

Pratikum 03 Insertion sort

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr

data3 = [12, 11, 13, 5, 6]
print("Output Insertion Sort :", insertion_sort(data3))
```

Penugasan 01 Eksperimen Bubble sort

```python
def bubble_sort_desc(arr):
    n = len(arr)
    for i in range(n-1):
        for j in range(n-i-1):
            # ubah tanda > menjadi < agar descending
            if arr[j] < arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

data = [64, 34, 25, 12, 22, 11, 90]
print("Descending Bubble Sort:", bubble_sort_desc(data))
```

Penugasan 02 Eksperimen Selection sort



```python
def selection_sort_max(arr):
    n = len(arr)
    for i in range(n-1, 0, -1):
        max_index = 0
        for j in range(1, i+1):
            if arr[j] > arr[max_index]:
                max_index = j
        arr[i], arr[max_index] = arr[max_index], arr[i]
    return arr

data = [64, 25, 12, 22, 11]
print("Selection Sort Max-to-End:", selection_sort_max(data))
```

Penugasan 03 Eksperimen Insertion Sort

```python
def insertion_sort_string(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j].lower() > key.lower():  # pakai lower untuk akurat
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr

data = ["Budi", "Andi", "Doni", "Cici"]
print("Insertion Sort String:", insertion_sort_string(data))
```

Penugasan 05 Analisis Kompleksitas

```python
import random

def count_bubble(arr):
    count=0
    n=len(arr)
    for i in range(n-1):
        for j in range(n-i-1):
            count+=1
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return count

def count_selection(arr):
    count=0
    n=len(arr)
    for i in range(n):
        min_idx=i
        for j in range(i+1,n):
            count+=1
            if arr[j]<arr[min_idx]:
                min_idx=j
        arr[i],arr[min_idx]=arr[min_idx],arr[i]
    return count

def count_insertion(arr):
    count=0
    for i in range(1,len(arr)):
        key=arr[i]
        j=i-1
        while j>=0 and arr[j]>key:
            count+=1
            arr[j+1]=arr[j]
            j-=1
        count+=1
        arr[j+1]=key
    return count

for n in [10,50,100]:
    data=[random.randint(1,1000) for _ in range(n)]
    print(f"\nUkuran input: {n}")
    print("Bubble Sort   :", count_bubble(data.copy()))
    print("Selection Sort:", count_selection(data.copy()))
    print("Insertion Sort:", count_insertion(data.copy()))
```

Penugasan 06 Proyek Mini

```python
def sort_barang(barang, by="nama", mode="asc"):
    reverse = True if mode=="desc" else False

    if by == "nama":
        return sorted(barang, key=lambda x: x[0].lower(), reverse=reverse)
    elif by == "harga":
        return sorted(barang, key=lambda x: x[1], reverse=reverse)

barang = [
    ("Buku",15000),
    ("Pensil",5000),
    ("Tas",75000),
    ("Pulpen",10000)
]

print("Sort Nama Asc:", sort_barang(barang,"nama","asc"))
print("Sort Nama Desc:", sort_barang(barang,"nama","desc"))
print("Sort Harga Asc:", sort_barang(barang,"harga","asc"))
print("Sort Harga Desc:", sort_barang(barang,"harga","desc"))
```
